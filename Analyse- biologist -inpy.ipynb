```python
import numpy as np
```

```python
heures = np.array([0, 1, 2, 3, 4, 5])
cellules = np.array([100, 150, 210, 300, 430, 600])
```

```python
croissance_moyenne = np.mean(np.diff(cellules))
print(croissance_moyenne)
```

```python
%pip install matplotlib
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(heures, cellules)
plt.title("Croissance des cellules")
plt.xlabel("Temps (heures)")
plt.ylabel("Nombre de cellules")
plt.grid()
plt.show()
```
